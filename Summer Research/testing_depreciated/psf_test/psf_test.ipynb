```python
from astropy.convolution import convolve_fft
from astropy.io import fits
from photutils.psf_matching import make_kernel
```

```python
# load data
img277 = fits.getdata('f277w_cutout.fits')
psf277 = fits.getdata('psf_F277W.fits')
psf444 = fits.getdata('psf_F444W.fits')

# normalize PSFs
psf277 /= psf277.sum()
psf444 /= psf444.sum()

# match kernel
kernel = make_kernel(psf277, psf444)

# convolve image
matched = convolve_fft(img277, kernel, preserve_nan=True)
fits.writeto('f277w_matched_to_f444w.fits', matched)
```
