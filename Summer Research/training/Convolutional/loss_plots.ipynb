```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
```

```python
cwd = Path.cwd()
model_dir = cwd / 'models'
```

```python
# Open the JSON file and load its contents into a dict
with open(cwd / 'loss.json', encoding='utf-8') as file:
    history = json.load(file)
```

```python
fig, axes = plt.subplots(
    nrows=len(history), ncols=2, figsize=(12, 4 * len(history)), squeeze=False
)

for i, fold in enumerate(history):
    fold_history = history[fold]
    x = range(len(fold_history['train_acc']))

    ax = axes[i, 0]
    ax.plot(x, fold_history['train_acc'], '-o', label='train_acc')
    ax.plot(x, fold_history['val_acc'], '-o', label='val_acc')
    ax.legend()

    ax = axes[i, 1]
    ax.plot(x, fold_history['train_loss'], '-o', label='train_loss')
    ax.plot(x, fold_history['val_loss'], '-o', label='val_loss')
    ax.legend()

    plt.tight_layout()
```
